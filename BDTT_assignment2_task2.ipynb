### data preprocessing
For this dataset , the CSV file had columns with no headings, To solve this a schema is defined with the headings for each column and the data type. The dataset is loaded into the dataframe "gm". For this project the games are rated on two actions, "play" which is playtime and "purchase" which indicates whether the game was purchased or not. This two actions will be the basis for our recommendation system and in the next step two dataframes are created each focusing on a specific action between the two. 

```python
from pyspark.sql.types import *
#Defining the schema and the column heading for the dataset
myschema = StructType([
    StructField("userId",IntegerType()),
    StructField("Game", StringType()),
    StructField("Action", StringType()),
    StructField("count", DoubleType())
])
#Loading the dataset into our dataframe with the schema created
gm = spark.read\
    .schema(myschema)\
    .csv("/Volumes/teaching/datasets/assignment_2/task_2/steam-200k.csv", header = True, inferSchema = True)
display(gm)
```

```python
#Two dataframes are created with one displaying only the purchases and the other only the play time
gm1 = gm.filter(gm.Action == "purchase")
pl1 = gm.filter(gm.Action == "play")
```

```python
display(gm1)
```

### ml flow initation and indexing
MLflow is initialised to track and log model training runs automatically. An experiment is created under a shared workspace path derived from the notebook name, allowing all ALS model metrics, parameters, and artifacts to be recorded and compared across runs. The experiment is named Gm_recommender. Preparation for the ALS model begins in cell 7, a string indexer is used to give unique integers to the games names since ALS algorithm requires numerical identifiers for items. The new dataframe replaces the game names with the game index.

```python
#Importing the Mlflow experiment, the auto logging and setting our path for where the experiments will be saved
import mlflow
import logging

logging.getLogger('mlflow').setLevel(logging.ERROR)
mlflow.pyspark.ml.autolog()
mlflow.set_tracking_uri("databricks")
mlflow.set_registry_uri("databricks-uc")

ctx = dbutils.notebook.entry_point.getDbutils().notebook().getContext()
nb = ctx.notebookPath().get().split('/')[-1]

experiment_name = f"/Shared/{nb}_Gm_recommender"

if mlflow.get_experiment_by_name(experiment_name) is None:
    mlflow.create_experiment(name=experiment_name)
mlflow.set_experiment(experiment_name)
```

```python
from pyspark.ml.feature import StringIndexer
#Using string indexer to create unique integers for the games
indexer = StringIndexer(inputCol="Game", outputCol="gameIndex")
gm2 = indexer.fit(gm1).transform(gm1)
display(gm2)
```

```python
#creating new dataframe with the gameIndex so it can used for ALS model
gm3 = gm2.select("userId","gameIndex", "count")
display(gm3)
```

### Building ALS model
The data is split into training and testing with a 70/30 split. The training data is placed in the trainingDF dataframe and the testing data in the testDF. The als model is initiated with set paramters(maxIter= 5, rank=10,regParam=0.1).egParam (Regularisation Parameter): Controls how much the model penalises large factor values to prevent overfitting.maxIter (Maximum Iterations): The number of times the ALS algorithm alternates between solving for user factors and item factors.rank: The number of latent (hidden) features used to represent each user and each game. The model trains using our training dataset `model = als.fit(trainingDF)` and then predictions of whether the game was purchased or not(0 or 1) are run on the testing dataset. This model is then evaluated using root mean square error. The error is shown to be 0.10818, which indicates high performance from the model.

```python
#Splitting the data into testing and training
(trainingDF, testDF) = gm3.randomSplit([0.7, 0.3],seed=100)
```

```python
from pyspark.ml.recommendation import ALS
#initiating the ALS model with the defined parameters
als = ALS(maxIter=5,rank=10,regParam=0.1, userCol="userId", itemCol="gameIndex", ratingCol="count", nonnegative=True, seed=100)
#run the model on the training dataset 
model = als.fit(trainingDF)
```

```python
#Run predictions using the ALS model that was trained on the training data, now it makes predictions on the testing dataset
predictions = model.transform(testDF).dropna()
predictions.show()
```

```python
from pyspark.ml.evaluation import RegressionEvaluator
#Evaluating the performance of the model by checking the root mean square error
evaluator = RegressionEvaluator(metricName = "rmse", labelCol = "count", predictionCol = "prediction")
rmse = evaluator.evaluate(predictions)
print("Root-mean-square error is %g" %rmse)
```

### Recommendation based on purchases
From the model, the games were able to be ranked to create a recommendation system for two use cases. The first use case shows 10 recommendations for each user and the second case shows the top 10 most purchased games. The most purchased games would be a good recommendation system for new users where there is no previous data to refer to. The first use case is important for recommended new games to old users on the platform `.join(user_recs2, on="gameIndex")"`


```python
from pyspark.sql import functions as F

# Get top 10 game recommendations for all users
user_recs = model.recommendForAllUsers(10)

# Explode recommendations and join with game names
user_recs2 = gm2.select("Game", "gameIndex").distinct()
user_recs_exploded = user_recs.select("userId", F.explode("recommendations").alias("rec"))
user_recs_named = user_recs_exploded.select(
    "userId",
    F.col("rec.gameIndex").alias("gameIndex"),
    F.col("rec.rating").alias("predicted_rating")
).join(user_recs2, on="gameIndex")

display(user_recs_named)
```

```python
Top10recs = gm2.select("Game", "gameIndex").distinct()
#Listing the top 10 games according to the number of purchases using our predictions
predictions.filter("prediction > 0.5") \
    .join(user_recs2, on="gameIndex") \
    .groupBy("Game").count() \
    .orderBy("count", ascending=False) \
    .limit(10).display()
```

### finetunning model
A Parameter grid is built and each paramter is given a set of values that will be tested and a grid search checks every combination of the values in the set. Each iteraction is evaluated and at the end we get the best combination and values for the parameters. 

```python
from pyspark.ml.tuning import CrossValidator, ParamGridBuilder
#running a parameter grid to find the best parameters to finetune our ALS model
paramGrid = ParamGridBuilder() \
    .addGrid(als.maxIter, [10, 15, 20]) \
    .addGrid(als.rank, [5, 15, 20]) \
    .addGrid(als.regParam, [0.001, 0.01, 0.05]) \
    .build()

crossval = CrossValidator(estimator=als,
                          estimatorParamMaps=paramGrid,
                          evaluator=evaluator,
                          numFolds=3,
                          seed=100)

gridsearchModel = crossval.fit(trainingDF)
```

```python
bestModel = gridsearchModel.bestModel
#Printing the results of the grid search to see which parameters are best
print("patamters for the best model:")

print("rank parameter: %g" %bestModel._java_obj.parent().getMaxIter())
print("rank parameter: %g" %bestModel.rank)
print("RegParam parameter: %g" %bestModel._java_obj.parent().getRegParam())
```

### recommendation based on play time and model evaluation
For our second ALS model, the data being used is the game playtime. The game names are replaced with the game index and split into training and testing data. The model parameters are defined and run on the testing data. The predictions are run on the testing data and evaluated like being using the root mean square error. In this model, it is clear that it struggles with the game time values. The error was high.
The ALS model is checked in the paramgrid and grid search so as to finetune and output more accurate predictions.


```python
#Using the string indexer on the playing time dataframe and splitting the data into training and testing
indexer = StringIndexer(inputCol="Game", outputCol="gameIndex")
pl2 = indexer.fit(pl1).transform(pl1)
pl3 = pl2.select("userId","gameIndex", "count")
display(pl3)
(training, test) = pl3.randomSplit([0.7, 0.3],seed=100)
```

```python
from pyspark.ml.recommendation import ALS
#Running the ALS model on the training dataset with the playing time as count
als2 = ALS(maxIter=5,rank=10,regParam=0.1, userCol="userId", itemCol="gameIndex", ratingCol="count", nonnegative=True, seed=100)

model2 = als2.fit(training)
predictions2 = model2.transform(test).dropna()
#Finding the root mean square error of the predictions
rmse2 = evaluator.evaluate(predictions2)
print("Root-mean-square error is %g" %rmse2)
```

```python
#Checking the predictions table to see how the predictor worked in comparison to the actual values
display(predictions2)
```

```python
from pyspark.ml.tuning import CrossValidator, ParamGridBuilder
#Doing a parameter grid to finetune our model with selected values
paramGrid2 = ParamGridBuilder() \
    .addGrid(als2.maxIter, [10, 15, 20]) \
    .addGrid(als2.rank, [5, 15, 20]) \
    .addGrid(als2.regParam, [0.001, 0.01, 0.05]) \
    .build()

crossval2 = CrossValidator(estimator=als2,
                          estimatorParamMaps=paramGrid2,
                          evaluator=evaluator,
                          numFolds=3,
                          seed=100)
#initiating a grid search to see which is the best combination of the paramaters
gridsearchModel2 = crossval2.fit(training)
bestModel2 = gridsearchModel2.bestModel

```

```python
#Printing the results from the grid search to see which parameters are best
print("parameters for the best model:")
print("maxIter parameter: %g" %bestModel2._java_obj.parent().getMaxIter())
print("rank parameter: %g" %bestModel2.rank)
print("RegParam parameter: %g" %bestModel2._java_obj.parent().getRegParam())
```

```python
from pyspark.ml.recommendation import ALS
#Initiating the ALS model with the new parameters that were got from the grid search
als3 = ALS(maxIter=10,rank=5,regParam=0.001, userCol="userId", itemCol="gameIndex", ratingCol="count", nonnegative=True, seed=100)

model3 = als3.fit(training)
predictions3 = model3.transform(test).dropna()

rmse3 = evaluator.evaluate(predictions3)
print("Root-mean-square error is %g" %rmse3)
```

### Recommendations from model
The recommendation system using the games with the highest predicted playing time and ranks the top 10 in terms of playing hours. A list of top 10 recommendations for each user is also created. The gameindex are mapped to their original game names and populate the table. This creates a more readable recommendation system.

```python
from pyspark.sql import functions as F

# Map gameIndex back to game names
game_lookup = pl2.select("Game", "gameIndex").distinct()

# Join predictions with game names, compute average predicted playtime, and rank
game_rankings = predictions2.join(game_lookup, on="gameIndex") \
    .groupBy("Game") \
    .agg(
        F.avg("prediction").alias("avg_predicted_playtime")
    ) \
    .orderBy(F.desc("avg_predicted_playtime"))\
        .limit(10)
    
#Displaying the top 10 games according to average playtime overall
display(game_rankings)
```

```python
# Generate top 10 game recommendations for each user based on predicted play time
user_recs_playtime = model2.recommendForAllUsers(10)

# Map gameIndex back to game names
user_recs_playtime_exploded = user_recs_playtime.select("userId", F.explode("recommendations").alias("rec"))
user_recs_playtime2 = user_recs_playtime_exploded.select(
    "userId",
    F.col("rec.gameIndex").alias("gameIndex")
).join(game_lookup, on="gameIndex")

# Show only userId and Game columns
display(user_recs_playtime2.select("userId", "Game"))
```
